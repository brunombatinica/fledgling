Take advantage of linear interpolation

$$ x_t = x_0(1-t) + x_1t$$

since we know the point of impact we can solve for time of impact and adjust trajectory

```python
# define boundaries
x_bounds = (-10,10)
y_bounds = (-10,10)

class particle():
  # a 2d particle for extremely simple simulations

  def __init__(self,p_0,v_0,a_0,r,x_bounds,y_bounds):
    # initial positions
    self.x = p_0[0]
    self.y = p_0[1]

    #particle size
    self.r = r

    #inital velocity and acceleration
    self.v = v_0
    self.a = a_0

    self.x_bounds = x_bounds
    self.y_bounds = y_bounds

  def __print__(self):
    print(f"p:{self.p}\n v:{self.v}, \n ke:{self.ke}")

  def move(self,dt):
    #update the particles position across a period dt (1/FPS)

    # update velocity first?
    # update velocity from acceleration
    self.v[0] += self.a[0]*dt
    self.v[1] += self.a[1]*dt

    # 1. calculate expected position
    x_test = self.x + self.v[0]*dt
    y_test = self.y + self.v[1]*dt

    # detect and handle collisions
    self.x, self.y = self.continuous_box_collision(self.x, x_test, self.y, y_test)

    return self.x, self.y

  def continuous_box_collision(self,x0,x1,y0,y1):
    #takes advantage of linear interpolation between x0,y0 and y0,y1
    #where tc is a value between [0,1] -

    collision = True
    while collision == True: #allows code to handle multiple wall collisions ?I dont think order matters
      if x1 == x0 or y1 == y0:
        break

      # print(self.x,self.y)
      # print(x0,x1,y0,y1)

      #figure out type of collision
      if x1 <= self.x_bounds[0]+self.r: #left wall
        #use collision point
        xc = self.x_bounds[0]+self.r

        #solve simulataneous equations
        tc = (xc - x0) / (x1 - x0)
        yc = y0 * (1 - tc) + y1 * tc

        #adjust trajectory
        # left wall
        x1 = (2*tc - 1) * (x1 - x0) + x0
        x0 = xc
        y1 = y1
        y0 = yc

        #adjust velocity
        self.v[0] = -self.v[0]

      elif x1 >= self.x_bounds[1]-self.r: #elif to prevent multiple changes at once

        # bp()

        #use collision point
        xc = self.x_bounds[1]-self.r

        #solve simulataneous equations
        tc = (xc - x0) / (x1 - x0)
        yc = y0 * (1 - tc) + y1 * tc

        #adjust trajectory
        # left wall
        x1 = (2*tc - 1) * (x1 - x0) + x0
        x0 = xc
        y1 = y1
        y0 = yc

        #adjust velocity
        self.v[0] = -self.v[0]

      elif y1 <= self.y_bounds[0]+self.r:
        #use collision point
        yc = self.y_bounds[0]+self.r

        #solve simulataneous equations
        tc = (yc - y0) / (y1 - y0)
        xc = x0 * (1 - tc) + x1 * tc

        #adjust trajectory
        y1 = (2*tc - 1) * (y1 - y0) + y0
        y0 = yc
        x1 = x1
        x0 = xc

        #adjust velocity
        self.v[1] = -self.v[1]

      elif y1 >= self.y_bounds[1]-self.r:

        #bp()

        #use collision point
        yc = self.y_bounds[1]-self.r

        #solve simulataneous equations
        tc = (yc - y0) / (y1 - y0)
        xc = x0 * (1 - tc) + x1 * tc

        #adjust trajectory
        y1 = (2*tc - 1) * (y1 - y0) + y0
        y0 = yc
        x1 = x1
        x0 = xc

        #adjust velocity
        self.v[1] = -self.v[1]
      else:
        collision = False

    return x1,y1

```

```python
def elastic_collision(m1,v1,m2,v2):
  #calculate the resultant velocitied from a COM reference frame after a elastic collision

  #conservation of momentum and energy - system of equations with 2 solutions

  # m1*v1 + m1*v2 = m1*v_1 + m2*v_2
  # 0.5*m1*v1**2 + 0.5*m2*v2**2 = 0.5*m1*v_1**2 + 0.5*m2*v_2**2
  # this second part simplifies to v1 + v_1 = v2 + v_2 which is v_1 - v_2 = v2 - v1
  # this second equation is why the trivial solution is no longer valid - the velocities cannot stay the same

  #bp()

  A = np.array([[m1,m2],[1,-1]])
  b = [m1*v1 + m2*v2, v2-v1]
  x = np.linalg.solve(A,b)

  # print(A)

  return x[0],x[1]

# sanity check
# m1 = 6
# v1 = 0.13
# m2 = 0.88
# v2 = 2.89
# v_1,v_2 = elastic_collision(m1,v1,m2,v2)

# print(m1*v1 + m2*v2, m1*v_1 + m2*v_2)
# print(0.5*m1*v1**2 + 0.5*m2*v2**2,0.5*m1*v_1**2 + 0.5*m2*v_2**2)
```

```python
def m_to_r(m):
  #converts a mass to a radius for a unit density circle so A = m
  # m = pi * r**2
  # sqrt(m/np.pi) = r
  return np.sqrt(m/np.pi)
```

```python
# define boundaries
x_bounds = (-10,10)
y_bounds = (-10,10)

class particle():
  # a 2d particle for extremely simple simulations
  # rewrite to use np arrays
  # and ignore acceleration

  def __init__(self,p_0,v_0, a0 = [0,0], r = [1], x_bounds = [10,-10],y_bounds = [-10,10]):


    #prior positions
    self.p_prior = np.array(p_0)

    # initial positions
    self.p = np.array(p_0)

    #particle size and mass
    self.r = r
    self.m = np.pi*r**2

    #inital velocity and acceleration
    self.v = np.array(v_0)

    self.x_bounds = np.array(x_bounds)
    self.y_bounds = np.array(y_bounds)

  def mv(self):
    return self.m * self.v

  def ke(self):
    return 0.5* self.m * (np.linalg.norm([self.v])**2)
  
  def __str__(self):
    return f"p:{self.p}\n v:{self.v}, \n ke:{self.ke()} \n mv: {self.mv()}"

  def move(self,dt):
    #update past position
    self.p_prior = self.p

    #update the particles position across a period dt (1/FPS)
    # 1. calculate expected position

    x_test,y_test = self.p + self.v*dt


    # detect and handle collisions
    self.p = self.continuous_box_collision(self.p[0],x_test, self.p[1], y_test)

  def continuous_box_collision(self,x0,x1,y0,y1):
    #takes advantage of linear interpolation between x0,y0 and y0,y1
    #where tc is a value between [0,1] -

    collision = True
    while collision == True: #allows code to handle multiple wall collisions ?I dont think order matters

      # print(self.x,self.y)
      # print(x0,x1,y0,y1)

      #figure out type of collision
      if x1 <= self.x_bounds[0]+self.r: #left wall
        if x1 == x0:
          break

        #use collision point
        xc = self.x_bounds[0]+self.r

        #solve simulataneous equations
        tc = (xc - x0) / (x1 - x0)
        yc = y0 * (1 - tc) + y1 * tc

        #adjust trajectory
        # left wall
        x1 = (2*tc - 1) * (x1 - x0) + x0
        x0 = xc
        y1 = y1
        y0 = yc

        #adjust velocity
        self.v[0] = -self.v[0]

      elif x1 >= self.x_bounds[1]-self.r: #elif to prevent multiple changes at once
        if x1 == x0:
          break
        # bp()

        #use collision point
        xc = self.x_bounds[1]-self.r

        #solve simulataneous equations
        tc = (xc - x0) / (x1 - x0)
        yc = y0 * (1 - tc) + y1 * tc

        #adjust trajectory
        # left wall
        x1 = (2*tc - 1) * (x1 - x0) + x0
        x0 = xc
        y1 = y1
        y0 = yc

        #adjust velocity
        self.v[0] = -self.v[0]

      elif y1 <= self.y_bounds[0]+self.r:
        if y1 == y0:
          break
        #use collision point
        yc = self.y_bounds[0]+self.r

        #solve simulataneous equations
        tc = (yc - y0) / (y1 - y0)
        xc = x0 * (1 - tc) + x1 * tc

        #adjust trajectory
        y1 = (2*tc - 1) * (y1 - y0) + y0
        y0 = yc
        x1 = x1
        x0 = xc

        #adjust velocity
        self.v[1] = -self.v[1]

      elif y1 >= self.y_bounds[1]-self.r:
        if y1 == y0:
          break
        #bp()

        #use collision point
        yc = self.y_bounds[1]-self.r

        #solve simulataneous equations
        tc = (yc - y0) / (y1 - y0)
        xc = x0 * (1 - tc) + x1 * tc

        #adjust trajectory
        y1 = (2*tc - 1) * (y1 - y0) + y0
        y0 = yc
        x1 = x1
        x0 = xc

        #adjust velocity
        self.v[1] = -self.v[1]
      else:
        collision = False

    return np.array([x1,y1])
```

```python
def particle_collision(p1,p2,dt):
  #alternate form using actualy maths hahahaha

  #detecting collision is super easy
    if np.linalg.norm(p1.p - p2.p) <= (p1.r + p2.r):

      #calculate exact point of impact using the trajectories
      #use linear interpolation - find time when the 2 lines are exactly p1.r + p2.r apart

      #solve for time when they are exactly r1 + r2 apart
      # https://physics.stackexchange.com/questions/221895/time-until-next-collision

      # don't need linear interpolation cause i have the velocity and there is no acceleration

      # making the coding a little neeter by writing it out - I dont think hte algebr will cancel much

      #to increast stability just make it register tc as t0 if this throws an error
      try:
        a = (p1.v[0] - p2.v[0]) + (p1.v[1] - p2.v[1])
        b = 2* ( (p1.p_prior[0] - p2.p_prior[0]) * (p1.v[0] - p2.v[0]) + (p1.p_prior[1] - p2.p_prior[1])*(p1.v[1] - p2.v[1]) )
        c = (p1.p_prior[0] - p2.p_prior[0])**2 + (p1.p_prior[1] - p2.p_prior[1]) ** 2 - (p1.r+p2.r)**2

        tc = ( -b - np.sqrt( b**2 - 4 * a * c) ) / ( 2 * a )

        if np.isnan(tc) or tc <= 0:  #is.na typo was trowing an error
          tc = 0
          
      except:
        tc = 0

      #positions of impact
      p1c = p1.p_prior + tc * p1.v
      p2c = p2.p_prior + tc * p2.v

      #print(tc,p1c,p2c)


      ### calculate new velocities

      # calculate contact/normal fector
      norm = p2c - p1c
      unorm = norm / np.linalg.norm(norm)
      utangent = np.array([-unorm[1],unorm[0]])

      #then it becomes a 1D equation so works :)
      #work within this reference frame and everything is good

      #project velocities onto the norm and tangent

      #tangent velocities will not change as no force in this direction
      t_v1 = np.dot(utangent, p1.v) * utangent
      t_v2 = np.dot(utangent, p2.v) * utangent

      #norm velocities are a 1d collision
      n_v1 = np.dot(unorm,  p1.v) * unorm
      n_v2 = np.dot(unorm, p2.v) * unorm

      # the norms can be treated like a 1d collision
      n_v1, n_v2 = elastic_collision(p1.m,n_v1,p2.m,n_v2)

      #convert vectors back to original coordinates/?vector space - not sure
      # if that is the right term
      p1.v = n_v1 + t_v1
      p2.v = n_v2 + t_v2


      ### calculate new position
      # p0 + tc * V0 + dt - tc * v1 = p1 - travel for some time along inital trajectory then some time under new trajectory
      # actually we know the exact point of impact so can simple use hte second portion
      # pc + (dt-tc)*V1
      p1.p_prior = p1.p
      p2.p_prior = p2.p

      p1.p = p1c + (dt-tc) * p1.v
      p2.p = p2c + (dt-tc) * p2.v
```
